# All About Qubits

### The Notation of Qubit

일반적인 컴퓨터(고전 컴퓨터)는 정보를 **비트**로 표현한다. 비트는 이진 값으로, 0 또는 1의 두 가지 상태만 가질 수 있다. <br>
이는 물리적으로 사물이 서로 다른 두 가지 상태에 있음을 의미한다. 예를 들어, 전자 부품을 통과하는 전압의 임계치를 기준으로 0또는 1로 구분하는 방법이 있다.

반면, 양자 컴퓨터는 일반적인 비트 대신 **큐비트**라고 불리는 특수한 비트를 사용한다.

### Mathematical representation of qubits

양자컴퓨팅의 수학적 프레임워크는 **선형대수**에 기반한다. 큐비트는 기본적으로 두 개의 요소를 가진 열 벡터로 나타낸다. <br>
가장 기본적인 두 가지의 상태는 비트의 0과 1에 정확히 대응하며, 다음과 같이 표현할 수 있다 :
$$
\text{qubit state 0} = \begin{pmatrix} 1 \\ 0 \end{pmatrix},
\quad
\text{qubit state 1} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}.
$$

양자역학에서는 이러한 상태들을 디랙 표기법(혹은 브라-켓 표기법)이라 부른다. <br/>
큐비트의 상태 벡터를 켓(ket)이라고 부르며, $\ket{\cdot}$으로 표기한다. 따라서 위의 두 상태는 다음과 같이 표현된다:

$$
\ket{0} = \begin{pmatrix} 1 \\ 0 \end{pmatrix},
\quad
\ket{1} = \begin{pmatrix} 0 \\ 1 \end{pmatrix}.
$$

0과 1 외에 임의의 상태를 표현할 때는 일반적으로 $\ket{\psi}$로 표현한다. <br/>
모든 켓에는 대응되는 브라(bra)가 존재한다. 브라는 행 벡터로, 켓에서 대응하는 요소의 켤레 전치이다. <br/>
브라는 $\bra{\cdot}$으로 표현하며, 0과 1의 브라는 다음과 같이 표현할 수 있다:

$$
\bra{0} = \begin{pmatrix} 1 & 0 \end{pmatrix},
\quad
\bra{1} = \begin{pmatrix} 0 & 1 \end{pmatrix}.
$$

두 큐비트 사이의 상태 내적은 한 큐비트 상태의 브라와 켓의 내적을 통해 계산된다. <br/>
이 때 $\braket{\cdot|\cdot}$로 표현을 간소화할 수 있다. 이를 브라-켓 표현식이라고 한다.

$$
\bra{0} \ket{1} = \braket{0|1} = \begin{pmatrix} 1 & 0 \end{pmatrix} \cdot \begin{pmatrix} 0 \\ 1 \end{pmatrix} = 1 \cdot 0 + 0 \cdot 1 = 0.
$$

여기서, $\bra{0}$와 $\ket{1}$을 결합할 때 세로선을 하나 없애면서 $\braket{0|1}$로 표현한 것을 알 수 있다.<br/>
또한 내적값이 0인 것을 통해 두 상태는 서로 직교한다는 것을 알 수 있다.

$\ket{0}$과 $\ket{1}$은 서로 직교하며 기저를 이루며, 이를 계산 기저라고 한다. 계산 기저는 상태 벡터의 길이가 1이 되도록 정규화되어 있다. <br/>
길이는 일반적인 2차원 벡터와 같이 자기 자신과의 내적을 취한 후 제곱근을 취하면 된다.
$$
\sqrt{\braket{1|1}} = \sqrt{\begin{pmatrix} 0 & 1 \end{pmatrix} \cdot \begin{pmatrix} 0 \\ 1 \end{pmatrix}} = \sqrt{0 \cdot 0 + 1 \cdot 1} = \sqrt{1} = 1.
$$

### Superposition

큐비트는 $\ket{0}$과 $\ket{1}$ 외에도 0과 1을 동시에 갖는 중첩(Superposition) 상태가 존재한다. <br/>
중첩 상태의 큐비트는 기저 상태들의 선형 조합으로 표현이 가능하다:
$$
\ket{\psi} = \alpha \ket{0} + \beta \ket{1} = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}.
$$

이때 $\alpha$와 $\beta$는 다음을 만족하는 복소수이다:
$$
\alpha\alpha^* + \beta\beta^* = 1.
$$
$^*$ 기호는 복소 켤레를 나타내며, $\alpha$와 $\beta$는 확률 진폭이라고 부른다.

이제 조금 더 복잡해진 상태에 대해 내적을 구해보자. 다음과 같은 두 상태가 있다고 가정한다:
$$
\ket{\psi} = \alpha\ket{0} + \beta\ket{1}, 
\quad
\ket{\phi} = \gamma\ket{0} + \delta\ket{1}.
$$

둘 사이의 내적 $\braket{\phi|\psi}$를 구하기 위해선, $\ket{\phi}$의 브라 값을 먼저 계산해야 한다. <br/>
큐비트가 중첩 상태에 있을 때, 각 기저 상태의 브라를 하나씩 취함으로써 브라를 계산할 수 있다. <br/>
이때 진폭의 켤레복소수를 취해야 하는 것을 주의해야 한다.
$$
\bra{\phi} = \gamma^*\bra{0} + \delta^*\bra{1} = \begin{pmatrix} \gamma^* & \delta^* \end{pmatrix}.
$$

결국 두 중첩 상태의 내적은 다음과 같이 계산된다:
$$
\braket{\phi|\psi} = \begin{pmatrix} \gamma^* & \delta^* \end{pmatrix} \begin{pmatrix} \alpha \\ \beta \end{pmatrix}.
$$

### Measurement outcome probabilities

양자컴퓨터에서 측정을 수행하면 확률에 따라 정보가 0 또는 1로 결정된다. <br/>
이때의 확률은 확률진폭 $\alpha$와 $\beta$에 의해 결정된다:

$$
\text{Prob(measure and observe }\ket{0}\text{)} = |\alpha|^2, \\
\text{Prob(measure and observe }\ket{1}\text{)} = |\beta|^2.
$$

$|\cdot|^2$는 모듈로 제곱이라 부르며, 어떤 수를 그 수의 켤레복소수로 곱하는 연산이다. 즉 $|\alpha|^2 = \alpha\alpha^*$ 이다. <br/>
측정 결과는 0또는 1로 두 가지밖에 없으므로, $|\alpha|^2 + |\beta|^2 = 1$이 성립해야 한다. <br/>
유효한 큐비트 상태를 위해서는 양자 상태를 정규화해야 한다.

### Operations on qubit states

상태 벡터 $\ket{\psi}$를 새로운 상태 벡터 $\ket{\psi'}$로 수정하기 위해선, 벡터를 수정하는 수학적 방법이 필요하다.

$$
\ket{\psi} = \alpha\ket{0} + \beta\ket{1} \rarr \ket{\psi'} = \alpha'\ket{0} + \beta'\ket{1}.
$$

이때 2차원 벡터를 다른 2차원 벡터로 변환하기 위해서는 $2 \times 2$ 행렬 $U$를 곱해주면 된다.
$$
\ket{\psi'} = U\ket{\psi}.
$$
그러나 아무 행렬이나 사용할 수는 없다. 연산 후에도 측정 결과 확률의 합이 1이 되도록 각 상태의 정규화를 유지해야 한다. <br/>
즉, $|\alpha'|^2 + |\beta'|^2 = 1$을 만족해야 한다.

양자 상태의 길이를 보존하는 행렬을 통해 정규화를 유지할 수 있다. 이 행렬을 유니터리 행렬이라고 부르며, 수학적으로는 $UU^\dagger = I$인 행렬이다. <br/>
$^\dagger$는 전치 행렬의 모든 원소의 켤레복소수를 취하는 것을 의미하며, $I$는 $2\times 2$인 단위 행렬을 의미한다.

***Solution*** <br/>

```python
# Here are the vector representations of |0> and |1>, for convenience
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])


def normalize_state(alpha, beta):
    """Compute a normalized quantum state given arbitrary amplitudes.

    Args:
        alpha (complex): The amplitude associated with the |0> state.
        beta (complex): The amplitude associated with the |1> state.

    Returns:
        np.array[complex]: A vector (numpy array) with 2 elements that represents
        a normalized quantum state.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    # CREATE A VECTOR [a', b'] BASED ON alpha AND beta SUCH THAT |a'|^2 + |b'|^2 = 1
    norm = np.abs(alpha)**2 + np.abs(beta)**2
    alpha_prime = alpha / np.sqrt(norm)
    beta_prime = beta / np.sqrt(norm)

    # RETURN A VECTOR
    return alpha_prime * ket_0 + beta_prime * ket_1
```

```python
def inner_product(state_1, state_2):
    """Compute the inner product between two states.

    Args:
        state_1 (np.array[complex]): A normalized quantum state vector
        state_2 (np.array[complex]): A second normalized quantum state vector

    Returns:
        complex: The value of the inner product <state_1 | state_2>.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    # COMPUTE AND RETURN THE INNER PRODUCT
    bra_state_1 = np.conj(state_1)
    return np.dot(bra_state_1, state_2)


# Test your results with this code
ket_0 = np.array([1, 0])
ket_1 = np.array([0, 1])

print(f"<0|0> = {inner_product(ket_0, ket_0)}")
print(f"<0|1> = {inner_product(ket_0, ket_1)}")
print(f"<1|0> = {inner_product(ket_1, ket_0)}")
print(f"<1|1> = {inner_product(ket_1, ket_1)}")
```
