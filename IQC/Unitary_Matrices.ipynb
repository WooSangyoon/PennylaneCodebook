# Unitary Matrices

### Operations on Qubits

단일 큐비트 연산은 기존의 유효한 큐비트 상태를 다른 유효한 큐비트 상태로 변환하는 작업이다. <br/>
이는 $2 \times 2$ 크기의 행렬을 곱해 행렬-벡터 연산을 사용해 수행된다. <br/>
즉, 단일 큐비트 연산 $U$는 기존의 양자 상태 $\ket{\psi}$를 새로운 상태인 $\ket{\psi'}$로 변환한다.

$$
\ket{\psi} \to \ket{\psi'} = U\ket{\psi}.
$$

큐비트 상태 벡터는 정규화되어있고, 따라서 길이가 1이다. <br/>
즉 큐비트에 작용하는 모든 행렬은 이러한 속성을 항상 만족하는 행렬이어야 한다. 이러한 행렬을 **유니터리 행렬(Unitary matrix)** 이라고 한다. <br/>

더 엄밀히 말하면, $n \times n$ 복소수 행렬 $U$가 다음을 만족할 때 유니터리 행렬이라고 한다.

$$
UU^\dagger = U^\dagger U = I_n
$$

여기서 $I_n$은 $n$차원 단위 행렬을 의미한다. <br/>
또한, $U^\dagger$은 $U$의 켤레 전치(Conjugate transpose) 또는 수반 행렬(Adjoint)을 나타내는 표기법이다. 즉 행렬을 전치하고 각 원소를 그 원소의 복소켤레로 바꾼 것이다.

### Unitary Parametrization

$2 \times 2$ 복소 행렬을 정의할 때, 각 원소마다 실수부와 허수부가 있으므로 총 8개의 실수가 필요하다고 생각할 수 있다. <br/>
그러나 유니터리 행렬의 제약 조건으로 인해, 실제로는 3개의 실수만으로도 복소수 행렬을 정의할 수 있다. 이는 Exercise의 문제들을 통해 확인 가능하다.

### ***Exercise I.3.1*** $\;$
다음 행렬을 가정해 보자:

$$
U = \begin{pmatrix} a & b \\ c & d \end{pmatrix}.
$$

$a, b, c, d$는 복소수이다. 이때 행렬곱 $UU^\dagger$을 계산하고, $U$가 유니터리 행렬이 되기 위해 만족해야 하는 방정식들을 쓰시오. <br/>
이 결과들로부터 $U$의 행 벡터들에 대해 무엇을 알 수 있는가? <br/>
이제 $U^\dagger U$에 대해서도 계산을 수행하고, 그 결과가 $U$의 열 벡터에 대해 무엇을 말해 주는지 쓰시오.

***Solution*** <br/>

$U^\dagger = \begin{pmatrix} a^* & c^* \\ b^* & d \end{pmatrix}$ 이므로, $UU^\dagger$를 계산하면:

$$
UU^\dagger =  \begin{pmatrix} a & b \\ c & d \end{pmatrix} \begin{pmatrix} a^* & c^* \\ b^* & d^* \end{pmatrix}
= 
\begin{pmatrix} |a|^2+|b|^2 & ac^* + bd^* \\ a^*c + b^*d & |c|^2+|d|^2 \end{pmatrix}.
$$

이때 유니터리 행렬이 되기 위해선 $UU^\dagger = I$ 이어야 하므로, 각 원소를 정리하면:

$$
|a|^2+|b|^2 = 1, \\
ac^* + bd^* = 0, \\
a^*c + b^*d = 0, \\
|c|^2+|d|^2 = 1.
$$

$|a|^2+|b|^2 = 1$와 $|c|^2+|d|^2 = 1$을 통해, 두 행 벡터의 길이가 모두 1임을 알 수 있다. <br/>
또한 두 행의 곱은 모두 0이 나온 것을 통해 두 행이 서로 직교한다는 것을 알 수 있다. <br/>

$U^\dagger U$ 또한 비슷한 방식으로 열 벡터의 길이가 1임을 도출할 수 있다. <br/>
결국 $U$는 행과 열이 직교하고, 정규화되어있어야 유니터리 행렬이 될 수 있음을 의미한다.

### ***Exercise I.3.2*** $\;$
다음 행렬로부터 시작해 보자:

$$
U = \begin{pmatrix} a & b \\ c & d \end{pmatrix}.
$$

$a, b, c, d$는 복소수일 때, $U$는 다음과 같이 표현할 수 있음을 보여라.

$$
U = \begin{pmatrix} a & -e^{i\beta}c^* \\ c & e^{i\beta}a^* \end{pmatrix}.
$$

(이때 $\beta$는 실수이며, 복소수 $e^{i\beta}$는 위상을 의미한다.)

***Solution*** <br/>
유니터리 행렬은 $UU^\dagger = I$ 이므로, 양변에 적절히 역행렬을 취하면 $U^\dagger = U^{-1}$이 된다.
$U^\dagger$은:
$$
U^\dagger = \begin{pmatrix} a^* & c^* \\ b^* & d^* \end{pmatrix}.
$$
이고, 일반적인 $2 \times 2$ 행렬의 역행렬 공식에 따르면, $U^{-1}$는:
$$
U^{-1} = \frac{1}{\text{det}(U)} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}.
$$
이므로

$$
\begin{pmatrix} a^* & c^* \\ b^* & d^* \end{pmatrix} = \frac{1}{\text{det}(U)} \begin{pmatrix} d & -b \\ -c & a \end{pmatrix}.
$$

따라서 
$$
d = \text{det}(U) a^*, \\
b = -\text{det}(U) c^*.
$$

이를 다시 $U$에 옮겨 적으면, 다음과 같이 표현할 수 있다:
$$
U = \begin{pmatrix} a & -\text{det}(U) c^* \\ c & \text{det}(U) a^* \end{pmatrix}.
$$

$UU^\dagger = I$ 이므로, 양변에 행렬식을 취하면:
$$
\text{det}(UU^\dagger) = \text{det}(U)\text{det}(U^\dagger) = \text{det}(I) = 1.
$$

$U^\dagger = U^{-1}$이므로, 
$$
\text{det}(U) \text{det}(U^\dagger) = \text{det}(U) \text{det}(U^{-1}) = |\text{det}(U)|^2 = 1.
$$

따라서 $|\text{det}(U)| = 1$이고, 복소평면에서 절대값이 1인 복소수는 단위원 위에 있음을 의미하므로, <br/>
$|\text{det}(U)| = e^{i\beta}$와 같이 쓸 수 있다. <br/>

최종적으로 대입을 진행하면:
$$
U = \begin{pmatrix} a & -\text{det}(U) c^* \\ c & \text{det}(U) a^* \end{pmatrix} =  \begin{pmatrix} a & -e^{i\beta}c^* \\ c & e^{i\beta}a^* \end{pmatrix}.
$$

### ***Exercise I.3.3*** $\;$
이전 Exercise로부터 얻어진 결과를 이용하여, 임의의 유니터리 행렬이 다음과 같이 표현될 수 있음을 보여라.

$$
U(\phi, \theta, \omega)= 
\begin{pmatrix} 
e^{-i(\phi + \omega)/2}\cos(\theta/2) & -e^{i(\phi - \omega)/2}\sin(\theta/2) \\ 
e^{-i(\phi - \omega)/2}\sin(\theta/2) & e^{i(\phi + \omega)/2}\cos(\theta/2) 
\end{pmatrix}.
$$

$\phi, \theta, \omega$는 실수이다.

***Solution*** <br/>

먼저, 모든 복소수는 극형식으로 나타낼 수 있으므로,
$$
a = re^{i\alpha}, \quad c = se^{i\gamma}.
$$

라고 할 수 있다. 이때 $a$와 $c$가 직교하며, 정규화가 되어 있음을 이전 Exercise에서 확인했으므로, <br/>
$r = \cos(\theta/2),\; s = \sin(\theta/2)$ 로 설정할 수 있다. 해당 값을 U에 대입하면:

$$
U = \begin{pmatrix} a & -e^{i\beta}c^* \\ c & e^{i\beta}a^* \end{pmatrix}= \begin{pmatrix} e^{i\alpha}\cos(\theta/2) & -e^{i(\beta-\gamma)}\sin(\theta/2) \\ e^{i\gamma}\sin(\theta/2) & e^{i(\beta-\alpha)}\cos(\theta/2) \end{pmatrix}.
$$

여기서 모든 원소로부터 $e^{i\beta/2}$라는 공통되는 위상을 추출할 수 있다.
그러나 양자 상태 전체에 동일한 위상이 곱해지더라도 측정 확률은 변하지 않기 때문에, 이는 생략이 가능하다. 따라서 다시 적어보면:

$$
U = e^{i\beta/2} \begin{pmatrix} e^{i(\alpha-\frac{\beta}{2})}\cos(\theta/2) & -e^{i(\frac{\beta}{2}-\gamma)}\sin(\theta/2) \\ e^{i(\gamma - \frac{\beta}{2})}\sin(\theta/2) & e^{i(\frac{\beta}{2}-\alpha)}\cos(\theta/2) \end{pmatrix} 
= 
\begin{pmatrix} e^{-i(\beta-2\alpha)/2}\cos(\theta/2) & -e^{i(\beta-2\gamma)/2}\sin(\theta/2) \\ e^{-i(\beta-2\gamma)/2}\sin(\theta/2) & e^{i(\beta-2\alpha)/2}\cos(\theta/2) \end{pmatrix}.
$$

목표 형태는 $\phi, \theta, \omega$ 만을 사용하므로, $\alpha, \beta, \gamma$를 적절히 치환해야 한다. 목표 형태와 지수를 비교하면:

$$
\phi + \omega = \beta - 2\alpha, \\
\phi - \omega = \beta - 2\gamma. \\
$$

따라서 두 관계식을 통해

$$
\phi = \beta - \alpha - \gamma, \\
\omega = -\alpha + \gamma.
$$

를 도출해낼 수 있다. 해당 변수들을 대입하면, 처음 목표했던 형태의 행렬

$$
U(\phi, \theta, \omega)= 
\begin{pmatrix} 
e^{-i(\phi + \omega)/2}\cos(\theta/2) & -e^{i(\phi - \omega)/2}\sin(\theta/2) \\ 
e^{-i(\phi - \omega)/2}\sin(\theta/2) & e^{i(\phi + \omega)/2}\cos(\theta/2) 
\end{pmatrix}.
$$

를 얻을 수 있다.
