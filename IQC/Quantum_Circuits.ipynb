### Wires and Registers

### Gates and Operations

큐비트에 대한 연산 작업을 게이트(gate)라고 부른다. 게이트는 여러 종류가 있으며, 종류에 따라 적용되는 연산이 다르다.

<img src="./assets/QC_fig4.png" width="500"/>

양자 회로는 왼쪽에서 오른쪽으로 읽는다. 위 그림의 경우 가장 먼저 0번과 2번 큐비트에 삼각형 게이트를 적용하고, 마지막에 0번과 1번 큐비트에 각각 오각형과 원 게이트를 적용하는 방식이다.

또한 서로 다른 큐비트에 대한 연산은 병렬적으로 실행이 가능하다.

<img src="./assets/QC_fig5.png" width="900"/>

위 이미지에서 0번 큐비트의 오각형 게이트는 왼쪽으로 이동시켜 1, 2번 큐비트의 직사각형 게이트와 동시에 적용될 수 있다. <br/>
회로에 빈 공간이 존재할 경우, 게이트 간의 의존 관계가 유지되는 한 이러한 작업을 수행할 수 있다.

<img src="./assets/QC_fig6.png" width="500"/>

따라서, 양자 회로는 위와 같이 여러 층으로 이루어진 구조로 시각화할 수 있다.

### Circuit Depth

양자 회로는 알고리즘이기 때문에, 복잡성 및 자원을 고려하여 최대한 효율적으로 구현하는 것이 중요하다. 이때 가장 중요한 지표는 회로 깊이(circuit depth)이다. <br/>
회로 깊이는 가능한 한 병렬로 작업을 수행할 때, 회로가 실행되는 데 걸리는 시간 단계 수를 의미한다. <br/>

회로의 깊이는 다음과 같이 게이트를 하나의 블록으로 생각하고 구조물을 만들었을 때의 깊이를 생각해 보면 된다. <br/>
깊이가 깊어 보이지만 병렬로 작업을 수행할 수 있다는 특성을 고려하면 다음 예시의 깊이는 6이 된다.

<img src="./assets/QC_fig7.png" width="600"/>

### Measurements

양자 회로의 마지막에서는 최소 하나 이상의 큐비트를 측정해야 한다. 회로에서 측정은 다이얼이 달린 박스의 형태로 표현한다.
측정 기호는 회로의 층처럼 보이지만 깊이 계산에는 포함되지 않는다.

### Quantum Circuits in Pennylane

Pennylane에서 양자 회로는 함수로 표현된다. 이는 기본적으로 파이썬 함수이지만, 하나 이상의 양자 연산을 수행해야 하고 하나 이상의 측정값을 반환해야 한다. <br/>
두 개의 큐비트로 구성된 양자 회로를 작성하는 의사 코드는 다음과 같이 적을 수 있다:

```
import pennylane as qp

qml = qp

def my_quantum_function(params):

    # Single-qubit operations with no input parameters
    qp.Gate1(wires=0)
    qp.Gate2(wires=1)

    # A single-qubit operation with an input parameter
    qp.Gate3(params[0], wires=0)

    # Two-qubit operation with no input parameter on wires 0 and 1
    qp.TwoQubitGate1(wires=[0, 1])
    
    # Two-qubit operation with an input parameter on wires 0 and 1
    qp.TwoQubitGate2(params[1], wires=[0, 1])

    # Return the result of a measurement
    return qp.Measurement(wires=[0, 1])
```

큐비트에 대한 연산은 한 줄씩 진행되는 것을 알 수 있다. 각 연산은 큐비트를 지정해 주어야 하며, 일부 연산은 추가적인 파라미터를 받는다.

다음 회로를 만드는 함수를 구상해 보자. 함수의 이름은 `my_first_circuit`이다:

<img src="./assets/QC_fig9.png" width="500"/>

첫 번째 $H$ 게이트는 `qp.hadamard`를 통해 구현되며, <br/>
두 번째 게이트는 $CNOT$이라 불리는 2-큐비트 게이트이다. `qp.CNOT`으로 구현된다. <br/>
세 번째 게이트인 $RZ$는 `qp.RZ`이며, 해당 게이트는 실수 파라미터 `theta`에 의존한다. <br/>
마지막에는 `qp.probs`를 사용하여 확률을 반환할 수 있다. <br/><br/>


일련의 과정을 코드로 나타내면 다음과 같다:

```
def my_first_circuit(theta):

    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0,1])
    qp.RZ(theta, wires = 0)

    return qp.probs(wires = [0,1])
```

그러나 이것만으로는 회로를 실행시킬 수 없고, 추가적인 정보를 몇 가지 더 제공해 주어야 한다.

### Devices and QNodes

Pennylane에서 회로를 실행하고 작동시키기 위해서는 두 가지의 요소가 추가로 필요하다: <br/>
- 회로를 실행시키기 위한 **장치(device)**
- 회로를 디바이스에 바인딩하고 실행시키기 위한 **QNode**

Pennylane은 시뮬레이션 외에도 실제 양자 하드웨어에서 실행시킬 수 있도록 플러그인을 제공한다. <br/> 
Pennylane에서 장치를 구성하려면 장치의 이름과 큐비트(와이어)의 수를 사전에 알아야 한다.

```
dev = qp.device('device.name', wires=num_qubits)
```

표준 양자 시뮬레이터를 사용하려면 장치 이름에 `default.qubit`을 사용하면 된다. <br/>
또한 장치의 와이어에 문자열 레이블을 지정할 수 있다.

```
dev = qp.device('default.qubit', wires=["wires_a", "wires_b"])
```

장치가 준비되었다면, QNode를 구성할 수 있다. <br/>

<img src="./assets/QC_fig10.png" width="700"/>


QNode를 구성하는 방법은 두 가지가 있다. <br/>
첫 번째 방법은 `qp.QNode` 함수를 이용하는 것이다:

```
my_qnode = qp.QNode(my_circuit, my_device)
```

QNode가 생성되면, 해당 QNode를 구성하는 양자함수와 동일한 매개변수를 입력함으로써 함수처럼 호출이 가능해진다.


이전 섹션에서 다뤘던 `my_first_circuit`을 QNode로 변환해 보자:

```
dev = qp.device('default.qubit', wires = [0,1])

def my_first_circuit(theta):

    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0,1])
    qp.RZ(theta, wires = 0)

    return qp.probs(wires = [0,1])

my_first_QNode = qp.QNode(my_first_circuit, dev)
```


이후, `my_first_QNode(angle)`과 같이 회로의 파라미터를 통해 QNode를 호출하면, 해당 회로의 측정 확률을 얻을 수 있다.

두 번째 방법은 데코레이터(함수 바로 위에 `@...`)를 사용하는 것이다. <br/>

양자 함수에 `@qp.qnode(dev)`와 같이 데코레이터를 적용하면, 함수와 동일한 이름을 가진 QNode가 자동으로 생성되어 `dev` 기기에서 실행이 가능하게 된다. <br/>

```
dev = qp.device('default.qubit', wires = [0,1])

@qp.qnode(dev)
def my_first_circuit(theta):

    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0,1])
    qp.RZ(theta, wires = 0)

    return qp.probs(wires = [0,1])
```

해당 방식으로 생성된 QNode는 함수의 이름과 같아지므로, `my_first_circuit(angle)`과 같이 실행이 가능해진다.

````python
# This creates a device with three wires on which PennyLane can run computations
dev = qp.device("default.qubit", wires=3)


def my_circuit(theta, phi, omega):

    ##################
    # YOUR CODE HERE #
    ##################

    # IMPLEMENT THE CIRCUIT BY ADDING THE GATES

    # Here are two examples, so you can see the format:
    # qp.CNOT(wires=[0, 1])
    # qp.RX(theta, wires=0)

    qp.RX(theta, wires=0)
    qp.RY(phi,   wires=1)
    qp.RZ(omega, wires=2)

    qp.CNOT(wires=[0, 1])
    qp.CNOT(wires=[1, 2])
    qp.CNOT(wires=[2, 0])

    return qp.probs(wires=[0, 1, 2])


# This creates a QNode, binding the function and device
my_qnode = qp.QNode(my_circuit, dev)

# We set up some values for the input parameters
theta, phi, omega = 0.1, 0.2, 0.3

# Now we can execute the QNode by calling it like we would a regular function
my_qnode(theta, phi, omega)
````
