# Measurements

### Projective Measurements

실제 문제를 해결하기 위해서는 큐비트를 측정하고 결과를 해석하는 방법이 필요하다. <br/>

먼저, 다음과 같은 상태의 큐비트가 있다고 가정한다:

$$
\ket{\psi} = \alpha\ket{0} + \beta\ket{1}.
$$

이 큐비트를 측정하면 <br/>
상태 $\ket{0}$이 관측될 확률은 $|\alpha|^2 = \alpha\alpha^*$, <br/>
상태 $\ket{1}$이 관측될 확률은 $|\beta|^2 = \beta\beta^*$ 이다. <br/>

측정 결과의 확률을 더 엄밀하게 표현하려면 내적을 사용할 수 있다. <br/>
어떤 양자 상태 $\ket{\psi}$가 주어졌을 때, $\ket{\varphi}$를 포함하는 기저에 대해 측정을 수행하여 상태 $\ket{\varphi}$를 관측할 확률은 다음과 같다:

$$
\text{Pr}(\varphi) = |\braket{\varphi | \psi}|^2.
$$

이런 유형의 측정은 투영 측정(projective measurement)이라 하며, 각 기저벡터가 주어진 상태에 얼마나 기여하는가를 내포한다. <br/>
이 이름은 두 상태 벡터의 내적을 통해 두 벡터의 중첩을 계산한다는 사실로부터 유래되었으며, 선형대수의 투영과 유사한 개념을 갖는다.

### Bases

지금까지는 계산 기저로 표현된 상태($\ket{0}, \ket{1}$의 선형결합)의 측정 결과 확률에 대해 논의했다. <br/>
그러나 시스템이 $\ket{0}$인지 $\ket{1}$인지 알아내는 것이 유효하지 않는 경우도 존재한다. 
실제로 계산 기저로 측정하면 서로 다른 양자 상태를 구별하는 것이 불가능한 경우가 존재한다.(Ex. $\ket{+}$와 $\ket{-}$ ) <br/>

양자컴퓨터의 측정은 어떠한 기저를 기준으로 수행되며, 이 기저에 따라 측정 결과가 달라질 수 있다.

### Changing Things Up-Basis Rotations

다음과 같이 계산 기저로 이루어진 상태가 있다고 가정해 보자:
$$
\ket{\psi} = \alpha\ket{0} + \beta\ket{1}.
$$

그리고 이 상태를 Hadamard 기저로 바꾸고자 한다.

$$
\ket{0} = \frac{1}{\sqrt{2}}(\ket{+} + \ket{-}), \\
\ket{1} = \frac{1}{\sqrt{2}}(\ket{+} - \ket{-}).
$$

이므로, 이를 대입하여 Hadamard 기저로 표현하면:

$$
\ket{\psi} = \biggr( \frac{\alpha+\beta}{\sqrt{2}} \biggr)\ket{+} + \biggr( \frac{\alpha-\beta}{\sqrt{2}} \biggr)\ket{-}.
$$

이제 이 상태가 Hadamard 기저로 표현되어 있으므로, 해당 기저로 측정을 수행할 수 있다. 이 측정 또한 투영 측정이지만 기존과 다른 기저에서 측정하는 것 뿐이다.

<img src="./assets/M_fig2.png" width="400">

그러나 대부분의 실제 하드웨어에서는 계산 기저($\ket{0}, \ket{1}$) 외에 다른 기저에서의 측정이 쉽지 않다. <br/>
따라서 측정하고 싶은 기저를 계산 기저로 옮겨 놓은 뒤 계산 기저에 대한 측정을 진행함으로써 우회할 수 있다. 이를 기저 회전이라 한다. <br/>
(Ex. $\ket{+}$에 $H$를 적용: 계산 기저 $\ket{0}$에 대응)

$\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$에서, 내적을 사용해 $\text{Pr}(0)$과 $\text{Pr}(1)$을 계산하라.

***Solution*** <br/>

$\text{Pr}(0) = |\braket{0|\psi}|^2$ 이므로, $\braket{0|\psi}$를 먼저 계산하면: <br/>

$$
\braket{0|\psi} = \alpha\braket{0|0} + \beta\braket{0|1} = \alpha.
$$

따라서 $\text{Pr}(0) = |\alpha|^2 = \alpha^2$.

마찬가지로, $\text{Pr}(1) = \beta^2$.

### ***Exercise I.9.2*** $\;$
하나의 기저는 정규직교인 벡터로 구성되며, 직교 조건은 벡터들이 선형 독립임을 보장하므로 다른 어떤 큐비트 상태도 해당 기저를 통해 표현할 수 있다.

$$
\ket{+} = \frac{1}{\sqrt{2}} (\ket{0}+\ket{1}), \\
\ket{-} = \frac{1}{\sqrt{2}} (\ket{0}-\ket{1}).
$$

그렇다면 이 두 상태는 하나의 기저를 이루는가?

***Solution*** <br/>

기저를 이루는지 확인하기 위해선, 정규화 여부와 직교 여부를 확인하면 된다. <br/>
우선 정규화 여부를 판단한다. 이는 자기 자신의 내적을 통해 확인할 수 있다.

$$
\begin{align}
\braket{+|+} &= \frac{1}{2}(\bra{0}+\bra{1})(\ket{0}+\ket{1}) \\
             &= \frac{1}{2} (\braket{0|0} + \braket{0|1} + \braket{1|0} + \braket{1|1}) \\
             &= \frac{1}{2} (1+0+0+1) \\
             &= 1.
\end{align}
$$

같은 방식으로, $\braket{-|-} = 1$ 이다. 즉, 각 상태가 정규화되어있음을 알 수 있다. <br/>

다음으로, 두 상태의 직교 여부를 확인하면:

$$
\begin{align}
\braket{+|-} &= \frac{1}{2}(\bra{0}+\bra{1})(\ket{0}-\ket{1}) \\
             &= \frac{1}{2} (\braket{0|0} - \braket{0|1} + \braket{1|0} - \braket{1|1}) \\
             &= \frac{1}{2} (1-0+0-1) \\
             &= 0.
\end{align}
$$

따라서 두 상태는 직교한다. <br/>

최종적으로, 이 두 상태는 정규화와 직교 여부를 모두 만족하므로 하나의 기저를 이룬다.

### ***Exercise I.9.3*** $\;$

내적을 활용하여, $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$를 Hadamard 기저($\ket{+}, \ket{-}$)에서 측정했을 때의 확률을 구하라.


***Solution*** <br/>

$\ket{\psi}$를 Hadamard 기저로 표현하면 다음과 같다.

$$
\ket{\psi} =  \frac{\alpha+\beta}{\sqrt{2}} \ket{+} +  \frac{\alpha-\beta}{\sqrt{2}} \ket{-}.
$$

해당 기저에서 측정 확률을 계산하면:

$$
\begin{align}
\text{Pr}(+) &= |\braket{+|\psi}|^2 \\
             &= \biggr| \frac{\alpha+\beta}{\sqrt{2}} \braket{+|+} +  \frac{\alpha-\beta}{\sqrt{2}} \ket{+|-} \biggr|^2 \\
             &= \biggr| \frac{\alpha+\beta}{\sqrt{2}} \biggr|^2 \\
             &= \frac{|\alpha+\beta|^2}{2}.

\end{align}
$$

마찬가지로 $\text{Pr}(-)$ 또한:

$$
\begin{align}
\text{Pr}(-) &= |\braket{-|\psi}|^2 \\
             &= \biggr| \frac{\alpha+\beta}{\sqrt{2}} \braket{-|+} +  \frac{\alpha-\beta}{\sqrt{2}} \ket{-|-} \biggr|^2 \\
             &= \biggr| \frac{\alpha-\beta}{\sqrt{2}} \biggr|^2 \\
             &= \frac{|\alpha-\beta|^2}{2}.

\end{align}
$$
