# X and H

### Describing Quantum Operations

실제 게이트들을 살펴보기 전에, 연산이 어떻게 작용하는지를 간단히 나타내는 표기법을 아는 것이 좋다. <br/>
먼저, 어떤 연산 $U$를 가정한다. $U$가 두 계산 기저에 어떻게 작용하는지만 알면, 다른 임의의 상태에 대한 $U$의 작용도 간단히 계산할 수 있다. <br/>

$U$가 계산 기저에 다음과 같이 적용된다고 가정한다:

$$
U\ket{0} = \alpha\ket{0} + \beta\ket{1}, \\
U\ket{1} = \gamma\ket{0} + \delta\ket{1}.
$$

임의의 상태 $\ket{\psi} = p\ket{0} + q\ket{1}$에 대한 작용을 계산하려면, 행렬-벡터 곱이 선형적이라는 성질을 이용하면 된다.

$$
U\ket{\psi} = U(p\ket{0} + q\ket{1}) = U(p\ket{0}) + U(q\ket{1}) = pU\ket{0} + qU\ket{1}.
$$

### Pauli X

가장 먼저 다루는 게이트는 **파울리 X(Pauli X)** 게이트이다. 이 연산은 다음과 같이 적용된다:

$$
X\ket{0} = \ket{1}, \\
X\ket{1} = \ket{0}.
$$

이는 불리언 NOT 연산과 비슷하기 때문에, 비트 반전 연산 또는 NOT 연산이라고도 불린다. <br/>
파울리 X 게이트를 행렬로 표현하면 다음과 같이 나타난다:

$$
X = \begin{pmatrix} 0 & 1 \\ 1 & 0 \end{pmatrix}.
$$

양자 회로에서는 두 가지 방식으로 표현한다.

<img src="./assets/XH_fig1.png" width="700"/>

첫 번째 방식은 박스 안에 X를 기입하는 식으로, 일반적으로 사용하는 형태이다. <br/>
두 번째 방식은 원 안에 세로선이 그어져 있는 방식으로, 단일 큐비트에서는 거의 사용하지 않고 2큐비트 연산에서 사용된다.

### Hadamard

다음으로 살펴볼 게이트는 아다마르(Hadamard) 게이트로, 다음과 같이 나타난다:

$$
H = \frac{1}{\sqrt{2}} \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}
$$

또한 회로도에서는 다음과 같이 표현한다.

<img src="./assets/XH_fig2.png" width="400"/>

아다마르 게이트는 $\ket{0}$과 $\ket{1}$ 두 상태에 균일한 중첩을 만들어줄 수 있다.

$$
H\ket{0} = \frac{1}{\sqrt{2}}(\ket{0} + \ket{1}), \\
H\ket{1} = \frac{1}{\sqrt{2}}(\ket{0} - \ket{1}).
$$

또한, 이 상태는 자주 발생하므로 진폭의 부호에 따라 특별한 명칭이 붙여져 있다.

$$
\ket{+} = \frac{1}{\sqrt{2}}(\ket{0} + \ket{1}), \\
\ket{-} = \frac{1}{\sqrt{2}}(\ket{0} - \ket{1}).
$$

### ***Exercise I.4.2*** $\;$
$X$ 연산의 고유값과 고유벡터를 구하라. <br/>
또한 고유벡터를 계산 기저 켓들의 선형결합 형태로 표현하라.

***Solution*** <br/>
고유벡터의 정의에 따라, 다음을 만족해야 한다:
$$
X\ket{v} = \lambda\ket{v}.
$$

$\ket{v} = \begin{pmatrix} a \\ b \end{pmatrix}$ 라 하면,

$$
\begin{pmatrix} b \\ a \end{pmatrix} = \lambda \begin{pmatrix}a \\ b\end{pmatrix} = \begin{pmatrix} \lambda a \\ \lambda b\end{pmatrix}.
$$

따라서 $b = \lambda a, \; a = \lambda b$ 를 얻을 수 있다. <br/>
$b = \lambda a$를 $a = \lambda b$에 대입하면:

$$
a = \lambda (\lambda a) = \lambda^2 a.
$$

고유벡터는 영벡터가 아니므로, $a \neq 0$인 경우 $\lambda = 1$ 또는 $\lambda = -1$이 된다. <br/>




**Case 1**: $\lambda = 1$인 경우 $b = a$ 이므로, 벡터는:

$$
\begin{pmatrix} \lambda a \\ \lambda b\end{pmatrix} = \begin{pmatrix} a \\ a \end{pmatrix} = a\begin{pmatrix} 1 \\ 1 \end{pmatrix} .
$$

따라서, 대표로 $\begin{pmatrix} 1 \\ 1 \end{pmatrix}$을 선택하여 정규화하면:

$$
\frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ 1 \end{pmatrix} = \frac{\ket{0} + \ket{1}}{\sqrt{2}} .
$$
<br/>


**Case 2**: $\lambda = -1$인 경우 $b = -a$ 이므로, 벡터는:

$$
\begin{pmatrix} \lambda a \\ \lambda b\end{pmatrix} = \begin{pmatrix} a \\ -a \end{pmatrix} = a\begin{pmatrix} 1 \\ -1 \end{pmatrix} .
$$

따라서, 대표로 $\begin{pmatrix} 1 \\ -1 \end{pmatrix}$을 선택하여 정규화하면:

$$
\frac{1}{\sqrt{2}} \begin{pmatrix} 1 \\ -1 \end{pmatrix} = \frac{\ket{0} - \ket{1}}{\sqrt{2}} .
$$

### ***Exercise I.4.3*** $\;$
입력 상태에 $HH\ket{0}$ 이나 $HH\ket{1}$과 같이 아다마르 게이트를 두 번 적용하면 어떻게 되는가? <br/>
브라켓 표기법을 사용해 결과를 계산하라.

***Solution*** <br/>
브라켓 표기법을 사용하면서 행렬-벡터 곱의 선형성을 이용하면, $HH\ket{0}$은:
$$
\begin{align}
HH\ket{0} &= H\biggr( \frac{1}{\sqrt{2}} (\ket{0} + \ket{1}) \biggr) \\
          &= \frac{1}{\sqrt{2}} H(\ket{0} + \ket{1}) \\
          &= \frac{1}{\sqrt{2}} \biggr( \biggr( \frac{1}{\sqrt{2}} (\ket{0} + \ket{1}) \biggr) + \biggr( \frac{1}{\sqrt{2}} (\ket{0} - \ket{1}) \biggr) \biggr) \\
          &= \frac{1}{\sqrt{2}} \biggr( \frac{2}{\sqrt{2}} \ket{0} \biggr) \\
          &= \ket{0}.
\end{align}
$$
<br/>

마찬가지로, $HH\ket{1}$은:
$$
\begin{align}
HH\ket{1} &= H\biggr( \frac{1}{\sqrt{2}} (\ket{0} - \ket{1}) \biggr) \\
          &= \frac{1}{\sqrt{2}} H(\ket{0} - \ket{1}) \\
          &= \frac{1}{\sqrt{2}} \biggr( \biggr( \frac{1}{\sqrt{2}} (\ket{0} + \ket{1}) \biggr) - \biggr( \frac{1}{\sqrt{2}} (\ket{0} - \ket{1}) \biggr) \biggr) \\
          &= \frac{1}{\sqrt{2}} \biggr( \frac{2}{\sqrt{2}} \ket{1} \biggr) \\
          &= \ket{1}.
\end{align}
$$

이를 통해, 아다마르 연산자는 자기 자신의 역함수이며, 균일한 중첩 상태를 생성하고 소멸시킬 수 있다는 것을 확인할 수 있다.

### ***Exercise I.4.4*** $\;$
$\ket{+}$와 $\ket{-}$가 함께 1큐비트 상태공간의 정규직교 기저를 이룬다는 것을 보여라. <br/>
즉, 정규화되어있고 직교함을 보여라 (이를 아다마르 기저라고 부른다).

***Solution*** <br/>
두 값이 모두 정규화되어있는지를 먼저 확인한다. 정규화 여부는 자기 자신을 내적했을 때의 값이 1이 나오는지 확인하면 된다.  <br/>
$\braket{+|+}$ 먼저 확인하면:

$$
\begin{align}

\braket{+|+} &= \biggr( \frac{\bra{0} + \bra{1}}{\sqrt{2}} \biggr) \biggr( \frac{\ket{0} + \ket{1}}{\sqrt{2}} \biggr) \\
             &= \frac{1}{2} (\bra{0} + \bra{1}) (\ket{0} + \ket{1}) \\
             &= \frac{1}{2} (\braket{0|0} + \braket{0|1} + \braket{1|0} + \braket{1|1}) \\
             &= \frac{1}{2} (1 + 0 + 0 + 1) \\
             &= 1.

\end{align} 
$$

동일한 방식으로, $\braket{-|-}$ 도 확인하면:
$$
\begin{align}

\braket{-|-} &= \biggr( \frac{\bra{0} - \bra{1}}{\sqrt{2}} \biggr) \biggr( \frac{\ket{0} - \ket{1}}{\sqrt{2}} \biggr) \\
             &= \frac{1}{2} (\bra{0} - \bra{1}) (\ket{0} - \ket{1}) \\
             &= \frac{1}{2} (\braket{0|0} - \braket{0|1} - \braket{1|0} + \braket{1|1}) \\
             &= \frac{1}{2} (1 - 0 - 0 + 1) \\
             &= 1.

\end{align} 
$$

이제 두 행렬이 직교하는지 확인하면:

$$
\begin{align}

\braket{-|+} &= \biggr( \frac{\bra{0} - \bra{1}}{\sqrt{2}} \biggr) \biggr( \frac{\ket{0} + \ket{1}}{\sqrt{2}} \biggr) \\
             &= \frac{1}{2} (\bra{0} - \bra{1}) (\ket{0} + \ket{1}) \\
             &= \frac{1}{2} (\braket{0|0} + \braket{0|1} - \braket{1|0} - \braket{1|1}) \\
             &= \frac{1}{2} (1 + 0 - 0 - 1) \\
             &= 0.

\end{align} 
$$

따라서, $\ket{+}$와 $\ket{-}$는 정규직교 기저를 이룬다.
