# It's Just a Phase

### Global and Relative Phases

임의의 양자 상태 $\ket{\psi}$를 다음과 같이 가정해 보자:

$$
\ket{\psi} = \alpha \ket{0} + \beta \ket{1}.
$$

이를 진폭의 실수 성분과 복소수 성분을 극좌표 형식으로 표현하면, 복소 위상 부분을 묶어서 표현할 수 있다.

$$
\ket{\psi} = ae^{i\theta}\ket{0} + be^{i\varphi}\ket{1} = e^{i\theta}(a\ket{0} + be^{i(\varphi-\theta)}\ket{1}).
$$

여기서 앞에 곱해진 $e^{i\theta}$ 항은 측정 결과 확률에 아무런 영향을 주지 않는다. <br/>
즉, 이 전역 위상을 무시할 수 있으며 동일한 양자 상태를 다음과 같이 표현할 수 있다:

$$
\ket{\psi} = a\ket{0} + be^{i(\varphi - \theta)}\ket{1} = a\ket{0} + be^{i\phi}\ket{1}.
$$

여기서 남아있는 복소수 인자 $e^{i\phi}$는 상대 위상이라고 하며, 이 또한 측정 확률에는 영향을 미치지 않는다는 것을 알 수 있다. <br/>
그러나, 이후 다른 측정 방식을 사용하게 되면 측정 결과에 영향을 미칠 수도 있음을 유의해야 한다.

### The Pauli Z Gate

$X$와 $H$ 게이트는 모두 중첩에 어떤 기저 상태가 포함되는지에 영향을 준다. <br/>
우리는 상태를 뒤집는 $X$ 게이트와, 균등한 중첩 상태를 만드는 $H$ 게이트에 대해 알고 있다. <br/>
본 절에서는 진폭을 보다 일반적인 방식으로 변경하려면 어떻게 해야 하는지를 살펴본다. 예를 들어, 상대 위상은 어떻게 바꾸는가에 대해 살펴본다.

먼저, Pauli $Z$ 게이트를 생각해 보면, 다음과 같이 작용한다:

$$
Z \ket{0} \to \ket{0}, \\
Z \ket{1} \to -\ket{1}.
$$

즉 $Z$ 게이트를 적용하면 $\ket{0}$은 그대로 유지되는 반면, $\ket{1}$ 상태는 위상이 뒤집히게 된다.

한편,

$$
Z\ket{+} = \frac{1}{\sqrt{2}}(\ket{0}-\ket{1}) = \ket{-}, \\
Z\ket{-} = \frac{1}{\sqrt{2}}(\ket{0}+\ket{1}) = \ket{+}.
$$

즉 $Z$ 게이트를 $\ket{+}$와 $\ket{-}$에 적용하면 두 성분 사이에 상대 위상이 발생하며 위상이 서로 뒤집히게 된다.

이는 수학적으로:

$$
Z = HXH.
$$

와 같다. (Codercise I.4.4에서 확인)

### Z Rotations

$RZ$ 게이트, 즉 $Z$ 회전($Z$ Rotation)은 Pauli $Z$보다 더 일반적인 방식으로 $\ket{0}$과 $\ket{1}$ 사이의 상대 위상을 변화시키는 게이트이다. <br/>
$RZ$ 게이트는 우리가 처음으로 접하는 매개변수화 게이트(Parametric Gate)로, 게이트의 동작이 어떠한 매개변수 $\omega$에 의해 결정된다. 이때 $\omega$는 라디안 각도의 단위이다.

예를 들어, 어떠한 초기 상태 $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$에 $\omega$ 만큼의 $Z$ 회전을 적용하면:

$$
RZ(\omega)\ket{\psi} = \alpha\ket{0} + \beta e^{i\omega}\ket{1}.
$$

$RZ$ 게이트를 행렬로 표현하면 다음과 같다:

$$
RZ = \begin{pmatrix} e^{-i\frac{\omega}{2}} & 0 \\ 0 & e^{i\frac{\omega}{2}} \end{pmatrix}.
$$

### S and T

Pauli $Z$ 게이트 외에도, $RZ$ 게이트에서 자주 사용되는 두 가지 경우가 존재한다. <br/>

첫 번째는 
$$
\omega = \pi / 2
$$
인 경우로, 이를 보통 Phase Gate라 부르며 $S$로 표기한다. <br/>

두 번째는 
$$
\omega = \pi / 4
$$
인 경우로, 이를 $T$ 게이트라고 한다. $T$ 게이트는 행렬 표현의 형태로 인해 “pi over 8 gate”라고도 불린다.

### ***Exercise I.5.1*** $\;$
임의의 상태 $\ket{\psi} = \alpha\ket{0} + \beta\ket{1}$에 대해 $RZ(\omega)$를 적용하라. <br/>
또한 그 결과가 아래 행렬 표현과 동일한 이유를 설명하라.

$$
RZ'(\omega) = \begin{pmatrix} 1 & 0 \\ 0 & e^{i\omega} \end{pmatrix}.
$$

***Solution*** <br/>

$RZ(\omega)\ket{\psi}$를 행렬로 표현하면:

$$
\begin{align}
RZ(\omega)\ket{\psi} &= \begin{pmatrix} e^{-i\frac{\omega}{2}} & 0 \\ 0 & e^{i\frac{\omega}{2}} \end{pmatrix} \begin{pmatrix} \alpha \\ \beta \end{pmatrix}
= \begin{pmatrix} e^{-i\frac{\omega}{2}}\alpha  \\  e^{i\frac{\omega}{2}}\beta \end{pmatrix} \\ 
&= e^{-i\frac{\omega}{2}}\alpha\ket{0} + e^{i\frac{\omega}{2}}\beta\ket{1}.
\end{align}
$$


이때 진폭의 복소수 부분을 따로 묶을 수 있다. 이는 전역 위상이므로 측정 확률 결과에 영향을 미치지 않으며, 따라서 생략 가능하다. 

$$
\begin{align}
e^{-i\frac{\omega}{2}}\alpha\ket{0} + e^{i\frac{\omega}{2}}\beta\ket{1} 
&= e^{-i\frac{\omega}{2}}(\alpha\ket{0} + e^{i\omega}\beta\ket{1}) \\
&= \alpha\ket{0} + e^{i\omega}\beta\ket{1} \\
&= \begin{pmatrix} 1 & 0 \\ 0 & e^{i\omega}\end{pmatrix} \begin{pmatrix} \alpha \\ \beta \end{pmatrix}.

\end{align}
$$

따라서 이러한 행렬 표현들이 서로 같다는 것을 보여준다. <br/>
(추가로, Pauli $Z$ 게이트는 $RZ$ 게이트에서 $\omega = \pi$일 때의 케이스이다.)

### ***Exercise I.5.2*** $\;$
$RZ(\omega)$의 역행렬(=켤레 전치행렬)을 $\omega$에 대한 함수로 표현하라.

***Solution*** <br/>

$RZ(\omega)$가 $\omega$만큼 회전을 가하는 것이므로, 반대방향으로 회전을 시켜 원위치로 복구할 수 있어야 한다. <br/>
즉,

$$
RZ^\dagger(\omega) = RZ(-\omega)
$$

로 표현할 수 있다.

### ***Exercise I.5.3*** $\;$
$S$의 행렬 표현을 구하라. 그리고 전역 위상 항을 제거한 후 고유값과 고유벡터를 계산하여라. <br/>
$S$를 두 번 적용하면 어떤 결과가 나타나는가?

***Solution*** <br/>

$S$는 $RZ(\pi/2)$를 의미하므로,

$$
S = \begin{pmatrix} e^{-i\frac{\pi}{4}} & 0 \\ 0 & e^{i\frac{\pi}{4}} \end{pmatrix} 
= e^{-i\frac{\pi}{4}} \begin{pmatrix} 1 & 0 \\ 0 & e^{i\frac{\pi}{2}} \end{pmatrix}.
$$

이때 $e^{-i\frac{\pi}{4}}$가 전역 위상이고, 오일러 공식에 의해 $e^{i\frac{\pi}{2}} = i$ 이므로

$$
e^{-i\frac{\pi}{4}} \begin{pmatrix} 1 & 0 \\ 0 & e^{i\frac{\pi}{2}} \end{pmatrix} 
= \begin{pmatrix} 1 & 0 \\ 0 & i \end{pmatrix}.
$$

이 게이트의 고유값은 1과 i이고, 각각에 대응하는 고유벡터는 $\ket{0}$과 $\ket{1}$이다. <br/>

$S$를 두 번 적용하면:

$$
|S|^2 = \begin{pmatrix} 1 & 0 \\ 0 & i\end{pmatrix} \begin{pmatrix} 1 & 0 \\ 0 & i\end{pmatrix} = \begin{pmatrix} 1 & 0 \\ 0 & -1\end{pmatrix}.
$$

이는 Pauli $Z$ 게이트와 같다. 따라서 $|S|^2 = Z$이다.
<br/>

### ***Exercise I.5.4*** $\;$
$S$ 게이트만을 사용해서 $S$의 수반 연산자(adjoint) $S^\dagger$를 구현하여라. <br/>
마찬가지로, $T$ 게이트만을 사용해서 $T^\dagger$를 구현하여라.

***Solution*** <br/>

Exercise I.5.3 에서 $S^2 = SS = Z$ 임을 확인하였다. 이때 $ZZ = I$이므로,
$$
ZZ = (SS)(SS) = I
$$
가 성립한다. <br/>

한편, $S^\dagger$는 $S$의 역연산이므로, 

$$
SS^\dagger = I
$$
가 성립한다. 따라서:

$$
SSSS = SS^\dagger = I
$$

즉 $S^\dagger = SSS$ 로 표현할 수 있다. <br/>


마찬가지로 $T$에서 $T^2 = S$이므로, 이를 $SSSS = I$에 대입하면:
$$
T^8 = TT^7 = I
$$
이고, 이 또한 $T^\dagger$가 T의 역연산이므로,

$T^\dagger = T^7$ 로 표현할 수 있다.

### I.5.1

```python
dev = qp.device("default.qubit", wires=1)


@qp.qnode(dev)
def apply_z_to_plus():
    """Write a circuit that applies PauliZ to the |+> state and returns
    the state.

    Returns:
        np.array[complex]: The state of the qubit after the operations.
    """

    ##################
    # YOUR CODE HERE #
    ##################

    # CREATE THE |+> STATE
    qp.Hadamard(wires=0)

    # APPLY PAULI Z
    qp.PauliZ(wires=0)
    # RETURN THE STATE

    return qp.state()


print(apply_z_to_plus())
```
