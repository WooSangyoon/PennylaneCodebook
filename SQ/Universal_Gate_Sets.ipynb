# Universal Gate Sets

### Universal Gate Sets

단일 큐비트의 가장 일반적인 유니터리 행렬은 다음과 같다:

$$
U(\phi, \theta, \omega) =
\begin{pmatrix}
e^{-i(\phi+\omega)/2}\cos(\theta/2) & -e^{i(\phi-\omega)/2}\sin(\theta/2) \\
e^{-i(\phi-\omega)/2}\sin(\theta/2) & e^{i(\phi+\omega)/2}\cos(\theta/2)
\end{pmatrix}
$$

그러나 사실 $U$는 조금 더 간단한 방법으로 표현할 수 있다. <br/>
모든 단일 큐비트 유니터리 연산은 두 가지 유형의 회전 게이트를 적절하게 사용하여 구현이 가능하다. <br/>
이러한 점은 양자 하드웨어 구축 측면에서 큰 장점이 될 수 있다.

$\{RX, RY, RZ \}$ 중 임의의 두 종류의 게이트만 있더라도 단일 큐비트 연산에 대한 범용 게이트 집합(Universal Gate Set)이 된다. <br/>
게이트 집합이 범용적(Universal)이라는 것은, 그 안에 포함된 게이트들을 조합하여 임의의 유니터리 행렬을 원하는 만큼 높은 정확도로 근사할 수 있다는 의미이다.

이를 더 형식적으로 말하면, 임의의 $U$에 대해 범용 게이트 집합에 속하는 게이트들만으로 구성된 $V$를 찾아서 <br/>

$$
||U-V|| \leq \epsilon
$$

을 만족시킬 수 있다. 이때 $\epsilon$은 임의로 작게 선택할 수 있다. (즉 몇 종류의 기본 게이트만으로도 원하는 만큼 $U$에 가깝게 만들 수 있다.) <br/>
이때 $V$를 구성하는 게이트를 찾아내는 것을 양자 회로 합성 (Quantum Circuit Synthesis)이라 한다.

$RX(\theta), RY(\theta), RZ(\theta)$ 같은 게이트를 사용하는 경우, $\theta$를 조절해 원하는 만큼 각도 회전이 가능하지만, <br/>
$H$, $T$ 같은 고정된 게이트들은 각도를 임의로 변경할 수 없다.  <br/>

그럼에도 불구하고, 임의의 단일 큐비트 연산은 $H$와 $T$만을 사용해서도 원하는 만큼 근사가 가능하다. <br/>
결국 각도가 고정된 게이트들만을 사용하더라도 게이트열이 길어질 수는 있겠으나 원하는 오차 $\epsilon$ 이내로 근사가 가능하다.

### ***Exercise I.7.1*** $\;$

$$
U(\phi, \theta, \omega) =
\begin{pmatrix}
e^{-i(\phi+\omega)/2}\cos(\theta/2)  &  -e^{i(\phi-\omega)/2}\sin(\theta/2) \\
e^{-i(\phi-\omega)/2}\sin(\theta/2)  &  e^{i(\phi+\omega)/2}\cos(\theta/2)
\end{pmatrix}.
$$

가 집합 $\{RZ, RY \}$에 속하는 게이트 3개만을 사용해 표현할 수 있음을 증명하라.


***Solution*** <br/>

$RZ$ 또는 $RY$가 연속으로 두 번 사용될 경우, 이는 두 회전각의 합만큼 회전하는 하나의 게이트로 볼 수 있다. <br/>
따라서 가능한 조합은 $RZ, RY, RZ$ 또는 $RY, RZ, RY$이다. <br/>

$$
RZ(\theta) = 
\begin{pmatrix}
e^{i\theta/2} & 0 \\ 0 & e^{i\theta/2}
\end{pmatrix}
, \;
RY(\theta) =
\begin{pmatrix}
\cos(\theta/2) & -\sin(\theta/2) \\ \sin(\theta/2) & \cos(\theta/2)
\end{pmatrix}.
$$

이고, 제시된 행렬에서 $\theta$에 의존하는 부분이 $RY$의 형태와 동일하므로:

$$
U(\phi, \theta, \omega) = RZ(\omega)RY(\theta)RZ(\phi).
$$

로 표현할 수 있다.

### ***Exercise I.7.2*** $\;$

$$
U(\phi, \theta, \omega) =
\begin{pmatrix}
e^{-i(\phi+\omega)/2}\cos(\theta/2)  &  -e^{i(\phi-\omega)/2}\sin(\theta/2) \\
e^{-i(\phi-\omega)/2}\sin(\theta/2)  &  e^{i(\phi+\omega)/2}\cos(\theta/2)
\end{pmatrix}.
$$

가 집합 $\{RZ, RX \}$에 속하는 게이트 3개만을 사용해 표현할 수 있음을 증명하라. <br/>
(hint: $Y = SXS^\dagger$ 이다.)

***Solution*** <br/>

$RY(\theta)$를 지수함수 형태로 표현하면:

$$
RY(\theta) = e^{-i\frac{\theta}{2}Y} = \cos\biggr(\frac{\theta}{2}\biggr)I - i\sin\biggr(\frac{\theta}{2}\biggr)Y.
$$

이때 $Y = SXS^\dagger$ 이므로:

$$
RY(\theta) = \cos\biggr(\frac{\theta}{2}\biggr)I - i\sin\biggr(\frac{\theta}{2}\biggr)SXS^\dagger.
$$

또한, $I = SIS^\dagger$미으로, 다음과 같이 정리할 수 있다:

$$
\cos\biggr(\frac{\theta}{2}\biggr)SIS^\dagger - i\sin\biggr(\frac{\theta}{2}\biggr)SXS^\dagger = S\biggr(\cos\biggr(\frac{\theta}{2}\biggr)I - i\sin\biggr(\frac{\theta}{2}\biggr)X\biggr)S^\dagger = SRX(\theta)S^\dagger.
$$


Exercise I.7.1에서 $U(\phi, \theta, \omega) = RZ(\omega)RY(\theta)RZ(\phi)$임을 확인했으므로, 이에 대입하면:

$$
U(\phi, \theta, \omega) = RZ(\omega)\biggr( SRX(\theta)S^\dagger\biggr) RZ(\phi).
$$

이때 $S = RZ(\pi/2), S^\dagger = RZ(-\pi/2)$이고, 동일한 축의 회전 게이트가 연속할 경우 이를 하나로 합칠 수 있으므로,

$$
RZ(\omega) RZ(\pi/2) RX(\theta) RZ(-\pi/2) RZ(\phi) = 
RZ \biggr(\omega + \frac{\pi}{2} \biggr) RX(\theta) RZ\biggr(\phi - \frac{\pi}{2}\biggr).
$$
