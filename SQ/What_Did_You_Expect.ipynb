### Observables

측정 결과의 확률을 구하면 큐비트 상태에 대한 유용한 정보를 얻을 수 있지만, 종종 에너지처럼 실제 물리량에 대응하는 값에 관심을 갖는 경우가 발생할 수 있다. <br/>
양자컴퓨팅에서는 이러한 물리량이 관측가능량(Observable)이라는 객체와 연결된다. <br/>

관측가능량은 에르미트 행렬(Hermitian Matrix)에 대응되며, 행렬의 고윳값들은 측정 결과로 나올 수 있는 가능한 값들을 나타낸다. <br/>
에르미트 행렬의 고윳값은 실수이며, 측정 가능한 물리량들은 실수여야 한다는 성질에 따라 자연스럽게 해석될 수 있다. <br/>


Pauli $Z$ 연산을 생각해 보면:
$$
Z = \begin{pmatrix} 1 & 0 \\ 0 & -1\end{pmatrix}.
$$

이 행렬은 에르미트이고, 고윳값은 $1$과 $-1$이며, 이에 대응하는 상태는 각각 $\ket{0}, \ket{1}$ 이다. <br/>

또한 측정 후에는 큐비트의 상태가 실제로 측정된 고윳값에 대응하는 고유벡터 상태로 바뀌게 된다. <br/>
만약 Pauli $Z$를 측정해서 $+1$이 나왔다면, 상태는 $\ket{0}$으로 붕괴하고, 결과가 $-1$이면 $\ket{1}$로 붕괴한다. <br/>

어떤 큐비트 상태에 대해 관측가능량의 값을 더 명확하게 파악하려면, 관측가능량의 기댓값을 측정해야 한다. <br/>
기댓값은 동일한 실험을 여러 번 반복했을 때 얻는 측정 결과들의 가중평균이다.  <br/>

예를 들어 Pauli $Z$ 측정에서 $+1$이 나올 확률이 $0.8$이고, $-1$이 나올 확률이 $0.2$라면, 기댓값은

$$
\braket{Z} = (+1)(0.8) + (-1)(0.2) = 0.6
$$
이다.

이를 수학적으로 표현하면, 상태 $\ket{\psi}$에 대해 어떤 관측가능량 $B$의 기댓값은 다음과 같다:
$$
\braket{B} = \braket{\psi|B|\psi}.
$$

### ***Exercise I.10.2*** $\;$

$$
\ket{\psi} = \frac{4}{5}\ket{0} - \frac{3}{5}e^{i\pi/3}\ket{1}.
$$
을 가정하자. 그리고 다음과 같은 관측가능량을 측정하고 싶다고 가정해 보자.

$$
B = \begin{pmatrix} 1 & -2i \\ 2i & 2 \end{pmatrix}.
$$

(1) 측정했을 때 나올 수 있는 가능한 결과는 무엇인가? <br/>
(2) 각 측정 결과에 대해, 측정이 끝난 뒤 큐비트가 어떤 상태인가? 이를 계산 기저로 표현하라. <br/>
(3) 관측가능량 $B$의 기댓값은 얼마인가? <br/>
(4) 이 상태를 준비한 뒤 $B$를 1000번 측정했다고 하자. 그 중 54번의 실험에서는 가능한 두 측정 결과 중 더 큰 값이 나왔고, 나머지 실험에서는 더 적은 값이 나왔다. 이때 실험적으로 얻은 $B$의 기댓값은 얼마인가? <br/>


***Solution*** <br/>

#### (1)
고윳값은 $\det(B-\lambda I)$를 풀어서 계산한다. <br/>
$$
\begin{align}
\det(B-\lambda I) &= \begin{pmatrix} 1-\lambda & -2i \\ 2i & 2-\lambda \end{pmatrix} \\
                  &= (1-\lambda)(2-\lambda) - (-2i)(2i) \\
                  &= \lambda^2 - 3\lambda -2 = 0.
\end{align}
$$

계산하면 $\lambda = \frac{3 \pm \sqrt{17}}{2}.$ <br/>
따라서, $\lambda_1 \approx 3.56155281$, $\lambda_2 \approx -0.56155281$.


#### (2) 
고유벡터를 $\ket{v} = \begin{pmatrix} a \\ b \end{pmatrix}$ 라 하자. 고유방정식이 $B\ket{v} = \lambda\ket{v}$ 이므로:

$$
B\ket{v} = \begin{pmatrix} 1 & -2i \\ 2i & 2 \end{pmatrix} \begin{pmatrix} a \\ b \end{pmatrix} = \lambda \begin{pmatrix} a \\ b \end{pmatrix} .
$$

첫번째 행에서, $a - 2ib = \lambda a$ 이므로, 이를 $b$에 대해 정리하면:

$$
b =\frac{i(\lambda-1)}{2}a.
$$

따라서,
$$
\ket{v_\lambda} \propto \ket{0} + \frac{i(\lambda-1)}{2}\ket{1}.
$$

해당 고유벡터에 (1)에서 구했던 $\lambda$를 대입하면:
$$
\ket{v_+} = \ket{0} + \frac{i(\frac{3 + \sqrt{17}}{2}-1)}{2}\ket{1}, \quad
\ket{v_-} = \ket{0} + \frac{i(\frac{3 - \sqrt{17}}{2}-1)}{2}\ket{1}. 
$$

두 벡터 모두 정규화 후 정리하면:

$$
\ket{v_1} = -0.61541221 i \ket{0} + 0.78820544 \ket{1}, \\
\ket{v_2} = 0.78820544 \ket{0} - 0.61541221 i \ket{1}.
$$


#### (3)
기댓값은 $\braket{\psi | B | \psi}$를 계산하여 얻을 수 있다.
$\ket{\psi} = \frac{1}{5} \begin{pmatrix} 4 \\ -3e^{i\pi/3} \end{pmatrix}$ 이므로:

$$
\braket{\psi | B | \psi} = \frac{1}{25} \begin{pmatrix} 4 & -3e^{i\pi/3} \end{pmatrix} \begin{pmatrix} 1 & -2i \\ 2i & 2 \end{pmatrix} \begin{pmatrix} 4 \\ -3e^{i\pi/3} \end{pmatrix} = -0.302769.
$$

#### (4)
실험으로 얻은 고유값을 계산하기 위해 가중 평균을 계산한다.

$$
\braket{\tilde{B}} = \frac{54*\lambda_1 + 946*\lambda_2}{1000} = -0.338905.
$$

```python
import numpy as np
import pennylane as qp
import matplotlib.pyplot as plt
```

```python
import matplotlib.pyplot as plt

def plotter(shot_vals, results_experiment, results_scaling):
    plt.figure(figsize=(10, 7))

    plt.scatter(
        shot_vals,
        results_experiment,
        facecolors="none",
        edgecolors="C0",
        s=80,
        linewidths=2,
        label="experiment"
    )

    plt.scatter(
        shot_vals,
        results_scaling,
        facecolors="none",
        edgecolors="C1",
        s=80,
        linewidths=2,
        label="function"
    )

    plt.xlabel("shots", fontsize=14, fontweight="bold")
    plt.ylabel("variance", fontsize=14, fontweight="bold")

    plt.xlim(0, max(shot_vals) * 1.05)
    plt.ylim(0, 0.10)

    plt.grid(True, alpha=0.4)

    plt.legend(
        title="var_type",
        loc="upper left",
        bbox_to_anchor=(1.02, 1)
    )

    plt.tight_layout()
    plt.show()
```
